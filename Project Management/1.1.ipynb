# 1. Тема лекции: Управление проектами в информационных технологиях

**Зачем нужно управление проектами в ИТ**  
Управление проектами применяют как лучший способ планировать и реализовывать ИТ‑инициативы, чтобы надёжно достигать целей и снижать затраты примерно на 10–15 %.  
Эта методология используется во всех крупных ИТ‑компаниях, поддерживается инструментами (например, Microsoft Project) и профессиональными ассоциациями вроде PMI.

**Что такое проект**  
- Проект – это временная работа, у которой есть начало и окончание, направленная на создание уникального продукта, услуги или результата.  
- «Временный» означает: проект заканчивается, когда цель достигнута или становится ясно, что её достичь нельзя.  
- «Уникальный» означает: результат отличается от уже существующих решений (новая система, внедрение ПО, новое оборудование и т.п.).

**Цели, критерии успеха и ограничения**  
- У проекта всегда есть одна или несколько целей: что именно нужно получить и каким путём.  
- Для оценки успеха нужны критерии: чаще всего сроки и стоимость, дополнительно учитываются качество и ресурсные ограничения.  
- Запланированные цели и качество выступают основными ограничениями, которые нельзя нарушать при поиске вариантов реализации.

**Чем управляет менеджер проекта**  
- Основные рычаги управления:  
  - технологии (какие методы и инструменты используются в проекте);  
  - ресурсы (люди, деньги, оборудование, материалы).  
- Вспомогательные рычаги:  
  - контракты (как привлекаются внешние ресурсы и подрядчики);  
  - организация работ (структура управления, каналы коммуникации, управление персоналом).  
- Информация в проектах всегда содержит неопределённость, поэтому важно учитывать риски и управлять ими.

**Группы процессов управления проектом (PMI PMBOK)**  
1. Инициация – принятие решения о начале проекта.  
2. Планирование – постановка целей, критериев успеха и разработка детального плана работ.  
3. Исполнение – координация команды и ресурсов для выполнения плана.  
4. Мониторинг и контроль – отслеживание хода работ, сравнение с планом, принятие и внедрение корректирующих действий.  
5. Закрытие – формальное завершение проекта и подведение итогов.

**Практические эффекты методологии управления проектами**  
- Обоснование целесообразности инвестиций в проект.  
- Формирование схемы финансирования работ.  
- Планирование сроков, ресурсов и затрат.  
- Организация взаимодействия участников проекта.  
- Планирование и управление качеством результатов.  
- Анализ и управление проектными рисками.  
- Планирование и управление контрактами.  
- Анализ отклонений от плана и прогноз их последствий.  
- Моделирование вариантов корректирующих действий и выбор управленческих решений.  
- Ведение архива проектов и использование накопленного опыта в новых проектах.

# 2. Основные термины и определения

## **2.1. Проект**

- Проект – это временная деятельность, направленная на создание уникального результата.  
  - Временная:  
    - есть начало и окончание;  
    - проект завершается, когда:  
      - цель достигнута;  
      - становится ясно, что цель недостижима;  
      - появляется решение остановить проект.  
  - Уникальная:  
    - результат отличается от уже существующих решений;  
    - примеры:  
      - разработка нового ПО;  
      - внедрение информационной системы;  
      - модернизация ИТ‑инфраструктуры.

## **2.2. Результат и цели проекта**

- Результат проекта:  
  - продукт (система, приложение, оборудование);  
  - услуга (поддержка, сопровождение, аутсорсинг);  
  - изменение состояния (улучшение процессов, сокращение затрат, повышение производительности).
- Цели проекта:  
  - описывают, что нужно получить и какую пользу это принесёт;  
  - включают не только конечный результат, но и выбранный путь (технологии, подходы, организацию работ).  
- Одна цель может достигаться разными вариантами решения, поэтому требуется выбор лучшего варианта по критериям.

## **2.3. Критерии успеха и ограничения**

- Основные критерии успеха:  
  - соблюдение сроков;  
  - укладывание в бюджет;  
  - достижение требуемого качества.  
- Ограничения проекта:  
  - зафиксированные цели и требования к результату;  
  - качество (стандарты, характеристики, метрики);  
  - ресурсы (люди, деньги, время, оборудование);  
  - внешние условия (законы, регламенты, политика компании).  
- Менеджер проекта постоянно балансирует между:  
  - сроками;  
  - стоимостью;  
  - качеством;  
  - объёмом работ (что включено в проект).

## **2.4. Ресурсы и технологии**

- Ресурсы проекта:  
  - человеческие ресурсы: разработчики, аналитики, тестировщики, админы, менеджеры;  
  - материальные ресурсы: серверы, рабочие станции, сети, помещения;  
  - нематериальные ресурсы: лицензии, ПО, доступы, документация;  
  - финансовые ресурсы: бюджет проекта.
- Технологии:  
  - методики разработки и внедрения (waterfall, agile, scrum, kanban и др.);  
  - технологический стек (языки, фреймворки, платформы);  
  - инструменты управления (Issue‑tracker, CI/CD, системы мониторинга).
- Управляя ресурсами и технологиями, менеджер проекта влияет на:  
  - сроки;  
  - стоимость;  
  - качество результата.

## **2.5. Неопределённость и риски**

- Неопределённость:  
  - неполная или неточная информация на старте;  
  - изменения требований по ходу проекта;  
  - внешние факторы (рынок, регулятор, технологии).
- Риск:  
  - возможное событие, которое может повлиять на проект;  
  - бывает негативным (угроза) и позитивным (возможность).
- Работа с рисками включает:  
  - выявление рисков;  
  - оценку вероятности и влияния;  
  - планирование реакций (избежать, снизить, передать, принять);  
  - мониторинг рисков в ходе проекта.

## **2.6. Процессы управления проектом (общая идея)**

- Управление проектом описывается как система процессов:  
  - процессы – это повторяемые типовые действия (анализ, планирование, контроль и т.д.);  
  - каждый процесс имеет входы (информация), действия (что делаем), выходы (документы, решения).
- Группы процессов (будут раскрыты дальше):  
  1. Инициация – старт проекта.  
  2. Планирование – разработка плана достижения целей.  
  3. Исполнение – реализация запланированных работ.  
  4. Мониторинг и контроль – отслеживание хода и корректировка.  
  5. Закрытие – формальное завершение проекта.
- Важно:  
  - процессы повторяются на разных фазах жизненного цикла;  
  - управление проектом – это не разовое действие, а постоянный цикл «спланировать → выполнить → проверить → скорректировать».


# 3. Основные положения управления проектами

## 3.1. Зачем организации запускают проекты

Организации используют проекты как инструмент реализации стратегии. Типичные причины запуска проекта:
- рыночная необходимость (ответ на изменение спроса, конкуренцию, новые требования клиентов);
- стратегическая возможность или бизнес‑потребность (создание нового продукта/сервиса для роста доходов);
- социальная потребность (проекты в сфере здравоохранения, образования, инфраструктуры);
- защита окружающей среды (экологические инициативы, «зелёные» технологии);
- требования заказчика или закона (контракты, новые нормативы);
- технологический прогресс (использование новых технологий для улучшения продукта или процессов).

## 3.2. Связь проектов со стратегией организации

- У организации есть стратегическое направление: цели, приоритеты, ключевые показатели.  
- Проекты должны поддерживать стратегию, а не существовать отдельно.  
- При изменении стратегии нужно пересматривать цели и содержание проектов.  
- Успешный проект приносит не только соблюдение сроков и бюджета, но и вклад в стратегические цели компании.

## 3.3. Принципы современной концепции управления проектами

1. Глубокое обоснование инвестиционных решений:
   - анализ нескольких вариантов проекта;
   - оценка технологических, экономических, социальных, экологических и других факторов.
2. Высокая координация и контроль:
   - прозрачные планы;
   - понятные роли и ответственность;
   - регулярный мониторинг выполнения работ.
3. Систематический учёт внешних изменений:
   - анализ рынка ресурсов и технологий;
   - учёт непредвиденных обстоятельств и негативных факторов;
   - своевременная корректировка планов.

## 3.4. Системный подход в управлении проектами

Системный подход предполагает:
- рассмотрение проекта как целостной системы, где элементы (задачи, люди, ресурсы, процессы) связаны между собой;
- понимание, что изменения в одной части затрагивают другие;
- декомпозицию проекта:
  - разбиение сложного проекта на этапы, подпроекты, пакеты работ;
  - упрощение планирования и контроля за счёт структурирования.

Преимущество системного подхода – возможность принимать взвешенные решения в сложных ситуациях, учитывая взаимосвязи и последствия.

## 3.5. Роль руководителя проекта в новой концепции

**Руководитель проекта (project manager):**
- несёт ответственность за достижение целей проекта в рамках ограничений (сроки, бюджет, качество, ресурсы);
- координирует работу команды и взаимодействие со всеми участниками.

**Основные задачи руководителя проекта:**
- определить полный состав работ и структурировать их (что делаем и в какой последовательности);
- сформировать и контролировать смету расходов;
- разработать и поддерживать актуальными графики работ;
- распределять ресурсы с учётом ограничений и рисков;
- обеспечивать качество работ и результатов;
- управлять рисками на всех этапах;
- выстраивать коммуникации с заказчиками, пользователями, руководством и другими заинтересованными сторонами.

## 3.6. Подходы к рассмотрению деятельности по управлению проектами

### 3.6.1. Функциональный подход

Фокусируется на управленческих функциях:
- анализ;
- планирование;
- организация;
- контроль;
- регулирование (корректировка).

Показывает, какие именно управленческие действия выполняет менеджер проекта.

### 3.6.2. Предметный подход

Структурирует управление по объектам, на которые направлено управление:
- объекты 1‑го типа – производственные (то, что создаётся: продукт, система, инфраструктура);
- объекты 2‑го типа – обеспечивающие элементы:
  - финансы;
  - кадры;
  - маркетинг;
  - риски;
  - материальные ресурсы;
  - качество;
  - информация и другие.

Помогает понять, какими объектами нужно управлять, чтобы проект был реализован.

### 3.6.3. Динамический подход

Рассматривает проект во времени, по стадиям:
- анализ проблемы;
- разработка концепции проекта;
- проектирование;
- строительство/разработка/внедрение;
- пусконаладка;
- эксплуатация;
- завершение проекта.

Показывает, как меняется содержание работ и управление на разных этапах жизненного цикла проекта.

### 3.6.4. Связь трёх подходов

- Функциональный подход – «что делает менеджер» (функции управления).  
- Предметный подход – «чем управляют» (объекты управления).  
- Динамический подход – «когда и в какой последовательности» (этапы во времени).  
Вместе они дают целостное понимание управления проектами.


# 4. Жизненный цикл проекта

## 4.1. Понятие жизненного цикла проекта

- Жизненный цикл проекта – это набор фаз, через которые проходит проект от инициации до закрытия.  
- Фазы, как правило:  
  - идут последовательно;  
  - имеют начало и конец (контрольные точки);  
  - задаются потребностями управления и контроля в конкретной организации.  
- Жизненный цикл проекта не зависит напрямую от жизненного цикла продукта, но проект должен учитывать, на какой стадии находится продукт (создание, развитие, сопровождение и т.д.).

## 4.2. Типичная структура жизненного цикла

Во всех проектах можно выделить базовую структуру жизненного цикла:

1. Начало проекта  
   - формулировка идеи;  
   - обоснование целесообразности;  
   - первичное определение целей и результатов.
2. Организация и подготовка  
   - уточнение требований;  
   - планирование работ, ресурсов, сроков и бюджета;  
   - формирование команды и организационной структуры.
3. Выполнение работ проекта  
   - реализация запланированных действий;  
   - создание продукта, услуги или результата;  
   - текущий контроль, взаимодействие участников.
4. Завершение проекта  
   - приёмка результатов;  
   - оформление завершения (отчёты, акты, документация);  
   - анализ опыта (что получилось хорошо, что улучшить).

Эта схема – «каркас» жизненного цикла, к которому можно привязывать более детальные модели и методологии.

## 4.3. Назначение жизненного цикла для управления

Жизненный цикл проекта помогает:

- структурировать большой объём работ по фазам, чтобы:  
  - на каждой фазе принимать решения «продолжать / остановить / скорректировать»;  
  - снижать неопределённость шаг за шагом;  
- увязать ключевые результаты (документы, версии продукта, решения) с контрольными точками;  
- эффективно организовать планирование и контроль:  
  - план работ и ресурсов можно уточнять от фазы к фазе;  
  - риски можно переоценивать на переходах между фазами.

## 4.4. Связь жизненного цикла и процессов управления

- Группы процессов управления (инициация, планирование, исполнение, мониторинг и контроль, закрытие) повторяются внутри каждой фазы жизненного цикла.  
- Пример:  
  - в начале фазы выполняется инициация этой фазы и её планирование;  
  - в середине – исполнение работ фазы и мониторинг;  
  - в конце – закрытие фазы и решение о переходе к следующей.  
- Поэтому нельзя путать:  
  - жизненный цикл (фазы проекта во времени);  
  - группы процессов управления (типовые управленческие действия, которые применяются на каждой фазе).

## 4.5. Жизненный цикл и неопределённость

- В начале проекта неопределённость максимальна, а информация – минимальна.  
- По мере прохождения фаз:  
  - появляется больше данных;  
  - уточняются требования, оценки сроков и затраты;  
  - часть рисков реализуется или снимается.  
- Разбиение на фазы позволяет:  
  - не «запирать» все решения на старте;  
  - принимать более информированные решения на переходах между фазами (go/kill/replan).


# 6. Формы организационной структуры

<span style="color:red; font-weight:bold">ВАЖНО:</span> структура организации сильно влияет на полномочия менеджера проекта, скорость коммуникаций и конфликты при распределении ресурсов.

## 6.1. Зачем понимать организационную структуру

- Проект реализуется внутри какой‑то организации (или нескольких), у которой уже есть своя структура управления.  
- От структуры зависят:
  - кому подчиняются люди, работающие в проекте;
  - какие у менеджера проекта есть полномочия;
  - как быстро согласуются решения и распределяются ресурсы.

**Три базовые формы:**
- функциональная структура;
- проектная структура;
- комбинированная (матричная) структура.

## 6.2. Функциональная организационная структура

- Организация разделена на функциональные подразделения:
  - разработка;
  - тестирование;
  - инфраструктура;
  - финансы;
  - маркетинг и т.п.
- Каждый сотрудник:
  - входит в своё функциональное подразделение;
  - имеет одного руководителя – функционального менеджера.

Особенности для проектов:
- Подразделения выполняют часть работ по проекту или весь проект в своей области.  
- Функциональный руководитель:
  - распределяет задачи и ресурсы;
  - решает, кого и на сколько времени выделить в проект.  
- Роль менеджера проекта:
  - часто размыта;
  - может быть «экспедитор»/«диспетчер проекта», который:
    - следит за сроками и обменом информацией;
    - но не имеет реальных полномочий по управлению людьми.

**Плюсы:**
- высокая функциональная специализация и профессионализм сотрудников;
- хорошая загрузка специалистов (меньше простоев).

**Минусы:**
- сложная интеграция работ между подразделениями;
- слабый фокус на проекте (приоритет – функциональные задачи);
- медленные согласования по иерархии;
- размытая ответственность за результат проекта.

## 6.3. Проектная организационная структура

- Вся команда проекта собирается в одном «проектном» подразделении:
  - разработчики, аналитики, тестировщики, админы и др. работают вместе;
  - их основная работа – именно этот проект.
- Менеджер проекта:
  - обладает максимальными полномочиями внутри проекта;
  - напрямую управляет командой;
  - отвечает за сроки, бюджет, качество и результаты.

Часто такая структура реализуется в виде проектного офиса.

**Плюсы:**
- сильная ориентация на цели проекта;
- эффективная коммуникация внутри команды;
- единое ответственное лицо за результат;
- проще интегрировать работы и контролировать проект.

**Минусы:**
- риск менее эффективного использования ресурсов:
  - специалисты «привязаны» к одному проекту;
  - возможны простои при неравномерной нагрузке;
- сложнее поддерживать развитие профессиональной специализации (меньше «функциональных» сообществ).

## 6.4. Комбинированная (матричная) организационная структура

- Попытка объединить плюсы функциональной и проектной структур.  
- Сотрудник имеет «двойное подчинение»:
  - по линии организации – функциональному руководителю;
  - по линии проекта – менеджеру проекта.

Типичный сценарий:
- функциональный руководитель:
  - отвечает за развитие специалистов, их квалификацию и распределение по проектам;
- менеджер проекта:
  - планирует работу по проекту;
  - согласует с функциональными руководителями участие людей;
  - управляет задачами, сроками и результатами внутри проекта.

**Плюсы:**
- лучшее использование ресурсов:
  - один специалист может участвовать в нескольких проектах;
  - поддерживается профессиональное развитие в функциональных «линиях»;
- можно дать менеджеру проекта достаточно полномочий, не ломая всю структуру компании.

**Минусы:**
- конфликты полномочий:
  - функциональный менеджер vs менеджер проекта;
- требования к организационной культуре:
  - нужны ясные правила распределения ответственности;
  - необходима готовность к сотрудничеству между менеджерами.

<span style="color:red; font-weight:bold">ЗАПОМНИ:</span> выбор структуры влияет на то, насколько реально у менеджера проекта есть власть исполнять свои обязанности, а не только «вести диаграммы и отчёты».

## 6.5. Когда какую структуру применять (интуитивно)

- Функциональная:
  - рутинная, повторяющаяся деятельность;
  - множество мелких задач, а не крупные уникальные проекты.
- Проектная:
  - крупные, критически важные проекты;
  - высокая цена ошибки и необходимость максимального фокуса.
- Комбинированная (матричная):
  - несколько параллельных проектов;
  - нужно балансировать между стабильной «функциональной» работой и проектной деятельностью.


# 7. Стандарты управления проектами

<span style="color:red; font-weight:bold">ВАЖНО:</span> стандарты не навязывают одну «правильную» методику, а дают общую рамку: термины, процессы, требования к компетенциям и качеству. На практике организации адаптируют стандарты под себя.

## 7.1. Виды стандартов управления проектами

По масштабу применения различают:

- Международные стандарты  
  - ориентированы на использование в разных странах;  
  - обычно включают не только описание процессов, но и обучение, сертификацию, аудит.
- Национальные стандарты  
  - действуют в пределах одной страны;  
  - учитывают национальное законодательство и практику.
- Общественные стандарты  
  - создаются профессиональными сообществами (ассоциациями);  
  - часто становятся де‑факто стандартом в отрасли.
- Частные стандарты  
  - разработаны отдельными компаниями или авторами;  
  - могут свободно распространяться.
- Корпоративные стандарты  
  - внутренние стандарты конкретной компании или группы компаний;  
  - обычно основываются на одном или нескольких международных/общественных стандартах.

## 7.2. PMI PMBOK – Project Management Body of Knowledge

<span style="color:red; font-weight:bold">КЛЮЧЕВОЙ СТАНДАРТ:</span> PMBOK – один из самых распространённых стандартов по управлению проектами.

Основные моменты:

- Разработан Институтом управления проектами (Project Management Institute, PMI).  
- Строится на процессном подходе:
  - управление описывается как совокупность процессов;
  - каждый процесс имеет входы, инструменты и выходы.
- Описывает области знаний по управлению проектами (классический набор):
  - управление интеграцией проекта;
  - управление содержанием (объёмом) проекта;
  - управление сроками;
  - управление стоимостью;
  - управление качеством;
  - управление человеческими ресурсами;
  - управление коммуникациями;
  - управление рисками;
  - управление закупками (контрактами).

Идея «пространства процессов»:

- Можно представить все процессы управления как точки в «пространстве», где осями являются:
  - фазы жизненного цикла;
  - группы процессов (инициация, планирование и др.);
  - области знаний (сроки, стоимость, качество и т.п.).
- Конкретные процедуры управления – это набор выбранных элементарных процессов, связанных между собой.

## 7.3. IPMA ICB – Competence Baseline

ICB – стандарт, который фокусируется не на процессах, а на компетенциях менеджера проекта.

Основные идеи:

- Разработан International Project Management Association (IPMA).  
- Определяет:
  - области компетентности (знания, умения, опыт) менеджеров проектов;
  - принципы оценки кандидатов для сертификации.
- Содержит:
  - 42 элемента компетенций;
  - из них 28 – основные, 14 – дополнительные.
- На основе ICB каждая страна – член IPMA – разрабатывает свой National Competence Baseline (NCB):
  - учитывает национальные особенности;
  - должен соответствовать требованиям IPMA и стандарту EN 45013 (сертификация).

<span style="color:red; font-weight:bold">Идея:</span> PMBOK отвечает «что делать и как», ICB – «каким должен быть менеджер, чтобы это делать».

## 7.4. ISO 10006 – управление качеством в проектах

ISO 10006 – международный стандарт по управлению качеством в проектах.

Основные положения:

- Фокус не только на качестве конечного продукта, но и на качестве процессов проекта.  
- Структура:
  - первая группа процессов – разработка стратегии (ориентация на потребности заказчика);  
  - вторая группа – управление взаимосвязями процессов;  
  - остальные группы – процессы, связанные с:
    - заданиями проекта;
    - сроками;
    - затратами;
    - ресурсами;
    - кадрами;
    - информационными потоками;
    - рисками;
    - закупками и материально‑техническим обеспечением.
- Ориентирован на проекты разных типов:
  - малые и крупные;
  - краткосрочные и долгосрочные;
  - технические, программные, сервисные и др.

Связь с другими стандартами качества:

- Использует терминологию ISO 8402 (общие определения по качеству).  
- Дополняет ISO 9004‑1 (качество продукции и процессов) фокусом на проектах.

## 7.5. Модели зрелости управления проектами (например, OPM3)

Организационная зрелость по управлению проектами:

- описывает способность организации:
  - выбирать правильные проекты;
  - управлять ими так, чтобы достигать стратегических целей.

Пример – стандарт PMI OPM3 (Organization Project Management Maturity Model):

- Состоит из трёх взаимосвязанных элементов:
  1. knowledge (знание)
     - база лучших практик по управлению проектами;
     - описание уровней зрелости.
  2. assessment (оценка)
     - инструменты для оценки текущего уровня зрелости;
     - выявление зон для улучшения.
  3. improvement (улучшение)
     - рекомендации, как переходить на более высокий уровень зрелости;
     - привязка улучшений к стратегическим целям компании.
- Задача таких моделей:
  - дать компании дорожную карту развития управления проектами;
  - повысить предсказуемость и успешность проектов.

## 7.6. Специфические методологии для ИТ‑проектов

Для ИТ‑проектов, помимо общих стандартов, применяются специальные модели и фреймворки, например:

- CMMI (Capability Maturity Model Integration)
  - модель зрелости процессов разработки ПО и систем;
  - помогает улучшать процессы анализа, проектирования, разработки, тестирования и сопровождения.
- Microsoft Solution Framework (MSF)
  - набор методик и моделей от Microsoft для управления ИТ‑проектами;
  - включает роли, процессы, подходы к управлению рисками и качеством.
- Rational Unified Process (RUP)
  - процесс разработки ПО, основанный на итерациях;
  - выделяет фазы: inception (инициация), elaboration (проработка), construction (создание), transition (внедрение);
  - задаёт артефакты, роли и практики для каждой фазы.

<span style="color:red; font-weight:bold">ЗАПОМНИ:</span> в реальных компаниях часто используют гибрид:  
общие стандарты (PMBOK/ICB/ISO) + корпоративные адаптации + ИТ‑специфические фреймворки (CMMI, RUP, agile‑подходы).


# 8. Процессы управления проектами

<span style="color:red; font-weight:bold">ВАЖНО:</span> процессы – это «повторяемые действия» менеджера проекта. Они идут не один раз, а циклически, в каждой фазе жизненного цикла.

## 8.1. Зачем нужны процессы управления

- Любой проект проходит через похожие управленческие шаги:
  - запуск идеи;
  - планирование;
  - выполнение;
  - контроль;
  - завершение.
- Процессы:
  - описывают эти шаги стандартизированно;
  - задают структуру: какие документы нужны, какие решения принимаются, какие действия выполняются.

<span style="color:red; font-weight:bold">Идея:</span> вместо хаотичного управления – повторяемый «конвейер» действий.

## 8.2. Связь процессов и жизненного цикла

- Жизненный цикл делит проект на фазы (начало, подготовка, выполнение, завершение).  
- В каждой фазе повторяются одни и те же группы процессов управления:
  - в начале фазы – инициация и планирование;
  - в середине – исполнение + мониторинг и контроль;
  - в конце – закрытие фазы.
- Это позволяет:
  - снижать неопределённость от фазы к фазе;
  - несколько раз пересматривать и уточнять планы.

## 8.3. Группы процессов управления проектом

Классический набор (по PMI PMBOK):

1. **Инициация**
   - Принятие решения «запускаем проект/фазу или нет».
   - Примеры действий:
     - формулировка цели и ожидаемых результатов;
     - предварительная оценка сроков и ресурсов;
     - подготовка и утверждение устава проекта.
2. **Планирование**
   - Преобразование идеи в конкретный план действий.
   - Примеры действий:
     - детализация требований;
     - декомпозиция работ (WBS – иерархическая структура работ);
     - оценка сроков и трудозатрат;
     - планирование ресурсов и бюджета;
     - планирование качества, рисков, коммуникаций, закупок.
3. **Исполнение**
   - Реализация запланированных работ.
   - Примеры действий:
     - постановка задач команде;
     - координация работ;
     - управление вовлечённостью стейкхолдеров;
     - обеспечение необходимых ресурсов.
4. **Мониторинг и контроль**
   - Сравнение фактического хода проекта с планом и корректировка.
   - Примеры действий:
     - сбор статусов и метрик (сроки, затраты, качество);
     - анализ отклонений;
     - принятие решений: что менять в плане или в исполнении;
     - управление изменениями (change requests).
5. **Закрытие**
   - Формальное завершение проекта или его фазы.
   - Примеры действий:
     - сдача результатов заказчику и получение приёмки;
     - закрытие контрактов;
     - подведение итогов и фиксация уроков (lessons learned);
     - архивирование документации.

## 8.4. Структура отдельного процесса

Каждый процесс можно представить как «чёрный ящик» со стандартной схемой:

- Входы:
  - документы (устав, план, требования, отчёты);
  - данные (метрики, статусы, риски).
- Инструменты и методы:
  - анализ, совещания, экспертные оценки;
  - программные инструменты (MS Project, Jira и др.);
  - техники (диаграммы Ганта, критический путь, матрицы рисков и т.п.).
- Выходы:
  - обновлённые планы;
  - отчёты и решения;
  - изменения в проекте (скорректированный бюджет, сроки, объём работ).

<span style="color:red; font-weight:bold">ЗАПОМНИ:</span> думай о процессе как о функции: «на вход – данные и документы, на выход – решение или новый документ».

## 8.5. Как процессы работают вместе

- Процессы связаны потоками информации:
  - результат одного процесса становится входом для другого;
  - пример:
    - анализ рисков (планирование) → план реагирования → действия по снижению рисков (исполнение) → мониторинг эффективности (контроль).
- Между процессами возможны циклы:
  - при серьёзных отклонениях мониторинг и контроль могут вернуть проект обратно к планированию;
  - при крупных изменениях требований – снова инициировать пересмотр устава и целей.

Так формируется «управленческий цикл»:
- спланировать → выполнить → проверить → скорректировать → повторить.

## 8.6. Практическое значение для менеджера проекта

Менеджеру проекта важно:

- знать, какие процессы существуют и в какой группе они находятся;
- понимать, какие входы/выходы у каждого процесса;
- выстраивать минимально достаточный набор процессов под конкретный проект:
  - маленький внутренний проект – облегчённый набор процессов;
  - большой сложный проект – более полный и формализованный набор.

<span style="color:red; font-weight:bold">КРАТКО:</span> процессы – это «инструкции», как управлять проектом шаг за шагом, а не набор красивых диаграмм ради отчётности.


# 9. Области знаний управления проектами (PMBOK)

<span style="color:red; font-weight:bold">ВАЖНО:</span> области знаний – это «что именно» нужно уметь контролировать в проекте: сроки, деньги, качество, риски и т.д. Процессы – «как управляем», области знаний – «чем управляем».

## 9.1. Общая идея областей знаний

PMBOK выделяет ключевые области, в которых менеджер проекта должен управлять:

1. Управление интеграцией проекта  
2. Управление содержанием (объёмом) проекта  
3. Управление сроками  
4. Управление стоимостью  
5. Управление качеством  
6. Управление человеческими ресурсами  
7. Управление коммуникациями  
8. Управление рисками  
9. Управление закупками (контрактами)

Каждая область включает набор процессов, распределённых по группам (инициация, планирование, исполнение, мониторинг и контроль, закрытие).

<span style="color:red; font-weight:bold">ЗАПОМНИ:</span> хорошее управление проектом = охватить все области знаний, а не только «сроки и бюджет».

## 9.2. Управление интеграцией проекта

**Цель:** собрать все части проекта в единую систему.

Основные моменты:
- разработка устава проекта;
- разработка и обновление интегрированного плана проекта;
- координация изменений:
  - любое изменение в сроках, стоимости, объёме или ресурсах согласуется и отражается в едином плане.

Пример: если увеличили объём работ, менеджер интеграции должен увязать это с дополнительными ресурсами, сроками и бюджетом, а не просто «дописать задачу в список».

## 9.3. Управление содержанием (объёмом) проекта

**Цель:** чётко определить, что входит в проект, а что – нет.

Основные задачи:
- сбор и анализ требований;
- формирование описания содержания проекта;
- создание WBS (Work Breakdown Structure – иерархической структуры работ):
  - разбиение проекта на пакеты работ;
  - привязка ответственности и оценок к пакетам;
- контроль изменений объёма (scope creep):
  - защита проекта от «тихого» расширения без ресурсов и времени.

<span style="color:red; font-weight:bold">ВАЖНО:</span> если область работ не зафиксирована, проект будет постоянно «раздуваться».

## 9.4. Управление сроками проекта

**Цель:** спланировать и контролировать календарь проекта.

Ключевые шаги:
- определение операций (конкретных задач);
- установление зависимостей между задачами;
- оценка длительности операций;
- разработка расписания (диаграммы Ганта, критический путь и т.п.);
- контроль сроков:
  - отслеживание фактического выполнения;
  - пересмотр расписания при отклонениях.

Результат – реалистичный график, с которым можно работать и который обновляется по мере выполнения проекта.

## 9.5. Управление стоимостью проекта

**Цель:** планировать и контролировать расходы проекта.

Основные элементы:
- оценка стоимости работ и ресурсов;
- формирование бюджета проекта:
  - распределение затрат по времени и этапам;
- контроль стоимости:
  - сравнение плановых и фактических затрат;
  - анализ отклонений и корректирующие действия.

Здесь часто применяется метод освоенного объёма (Earned Value Management), совмещающий анализ сроков и стоимости.

## 9.6. Управление качеством проекта

**Цель:** обеспечить соответствие продукта и процессов требованиям к качеству.

Состав:
- планирование качества:
  - выбор стандартов и метрик (дефекты, производительность, доступность и т.д.);
- обеспечение качества:
  - внедрение процессов, которые позволяют предотвращать брак;
- контроль качества:
  - проверки, тестирование, инспекции;
  - анализ несоответствий и улучшения.

Качество касается не только продукта (ПО, системы), но и процессов (как мы работаем).

## 9.7. Управление человеческими ресурсами проекта

**Цель:** сформировать и эффективно использовать команду проекта.

Основные направления:
- планирование ресурсов:
  - какие компетенции и роли нужны;
- набор и назначение команды:
  - подбор людей, распределение ролей и ответственности;
- развитие команды:
  - обучение, коучинг, обмен знаниями;
- управление командой:
  - мотивация;
  - разрешение конфликтов;
  - оценка эффективности работы.

В ИТ‑проектах это особенно критично: команда – главный актив.

## 9.8. Управление коммуникациями проекта

**Цель:** обеспечить правильный обмен информацией между всеми участниками.

Состав:
- планирование коммуникаций:
  - кто, какую информацию, в каком формате и как часто получает;
- распределение информации:
  - отчёты, митинги, рассылки, доски задач;
- мониторинг коммуникаций:
  - контроль, что информация доходит и понимается;
  - адаптация форматов (краткие summary, визуализации и т.д.).

<span style="color:red; font-weight:bold">ВАЖНО:</span> большинство проблем в проектах – не технические, а коммуникационные.

## 9.9. Управление рисками проекта

**Цель:** заблаговременно работать с неопределённостью, а не реагировать только при проблемах.

Ключевые шаги:
- планирование управления рисками (как именно будем с ними работать);
- идентификация рисков (мозговые штурмы, интервью, анализ похожих проектов);
- качественная и количественная оценка:
  - вероятность;
  - влияние на сроки, стоимость, качество;
- планирование реагирования:
  - избегание;
  - снижение;
  - перенос;
  - принятие;
- мониторинг и контроль рисков:
  - отслеживание триггеров;
  - актуализация реестра рисков.

Результат – осознанное управление «что может пойти не так» и «какими возможностями можно воспользоваться».

## 9.10. Управление закупками (контрактами) проекта

**Цель:** работать с внешними поставщиками так, чтобы они поддерживали, а не ломали проект.

Основные элементы:
- планирование закупок:
  - что выгоднее сделать внутри, а что – купить/заказать;
- выбор поставщиков:
  - запрос предложений, оценка, переговоры;
- администрирование контрактов:
  - контроль исполнения условий;
  - управление изменениями;
- закрытие контрактов:
  - приёмка поставок;
  - юридическое завершение договоров.

В ИТ‑проектах это могут быть:
- лицензии и подписки;
- оборудование и инфраструктура;
- услуги внешних подрядчиков (разработка, тестирование, интеграция).


# 10. Области знаний управления проектами (PMBOK)

<span style="color:red; font-weight:bold">ВАЖНО:</span> области знаний показывают, **чем** именно нужно управлять в проекте: не только сроками и деньгами, но и качеством, людьми, коммуникациями, рисками и стейкхолдерами.

## 10.1. Управление интеграцией проекта

**Цель:** собрать все части проекта в единую систему.

Основные задачи:
- разработка и утверждение устава проекта;
- создание и актуализация интегрированного плана проекта;
- управление изменениями:
  - любое изменение (сроки, бюджет, объём, ресурсы) согласуется и отражается в общем плане.

## 10.2. Управление содержанием (объёмом) проекта

**Цель:** чётко определить, что входит в проект, а что не входит.

Ключевые элементы:
- сбор и анализ требований;
- формирование описания содержания проекта;
- построение WBS (иерархической структуры работ);
- контроль изменений объёма (борьба с «ползущим» scope creep).

<span style="color:red; font-weight:bold">ВАЖНО:</span> плохо определённый объём почти гарантированно приводит к постоянным «доппросам» без ресурсов и времени.

## 10.3. Управление сроками проекта

**Цель:** спланировать и контролировать календарь проекта.

Основные шаги:
- определение перечня операций (задач);
- установление зависимостей между задачами;
- оценка длительности операций;
- построение расписания (диаграмма Ганта, критический путь и т.п.);
- контроль сроков и корректировка графика при отклонениях.

Результат – реалистичный и регулярно обновляемый график.

## 10.4. Управление стоимостью проекта

**Цель:** спланировать и контролировать расходы.

Состав:
- оценка стоимости работ и ресурсов;
- формирование бюджета (по этапам/фазам);
- контроль стоимости:
  - сравнение плановых и фактических затрат;
  - анализ отклонений и корректирующие действия.

Часто применяется метод освоенного объёма (EVM), совмещающий анализ сроков и бюджета.

## 10.5. Управление качеством проекта

**Цель:** обеспечить соответствие продукта и процессов установленным требованиям к качеству.

Основные компоненты:
- планирование качества (стандарты, метрики, критерии приемки);
- обеспечение качества (построение процессов, предотвращающих дефекты);
- контроль качества (проверки, тестирование, инспекции, анализ дефектов).

Качество относится и к результату (продукту), и к процессам проекта.

## 10.6. Управление человеческими ресурсами проекта

**Цель:** сформировать эффективную команду и поддерживать её работу.

Основные направления:
- планирование ресурсов: какие роли и компетенции нужны;
- набор и назначение команды;
- развитие команды (обучение, обмен знаниями, тимбилдинг);
- управление командой (мотивация, конфликты, оценка эффективности).

В ИТ‑проектах это критично: успех часто зависит не от инструментов, а от команды.

## 10.7. Управление коммуникациями проекта

**Цель:** обеспечить правильную и своевременную передачу информации между участниками проекта.

Ключевые действия:
- планирование коммуникаций:
  - кто, какую информацию, в каком формате и как часто получает;
- организация коммуникаций:
  - отчёты, митинги, чаты, доски задач;
- мониторинг и улучшение коммуникаций:
  - оценка, доходит ли информация и правильно ли понимается.

<span style="color:red; font-weight:bold">ВАЖНО:</span> большинство проблем в проектах возникают из‑за плохих коммуникаций, а не из‑за технологий.

## 10.8. Управление рисками проекта

**Цель:** работать с неопределённостью заранее, а не реагировать только на последствия.

Этапы:
- планирование управления рисками (как будем с ними работать);
- идентификация рисков;
- оценка вероятности и влияния;
- планирование реагирования:
  - избегание, снижение, перенос, принятие;
- мониторинг рисков и корректировка планов.

Результат – осознанный подход к тому, что может пойти не так (или лучше, чем ожидалось).

## 10.9. Управление закупками (контрактами) проекта

**Цель:** организовать работу с внешними поставщиками и подрядчиками.

Основные элементы:
- планирование закупок (что делать внутри, что покупать/аутсорсить);
- выбор поставщиков (запросы предложений, оценка, переговоры);
- администрирование контрактов (контроль сроков, качества, объёма);
- закрытие контрактов (приёмка, расчёты, юридическое закрытие).

В ИТ‑проектах это может включать: оборудование, лицензии, облачные сервисы, услуги подрядчиков.

## 10.10. Управление заинтересованными лицами проекта

**Цель:** учитывать интересы стейкхолдеров и управлять их влиянием на проект.

Основные шаги:
- выявление стейкхолдеров:
  - кто заинтересован в проекте;
  - кто может помочь или помешать;
- анализ интересов и ожиданий:
  - что каждый хочет получить;
  - какие есть опасения;
- планирование работы со стейкхолдерами:
  - кого и как вовлекать (эксперты, комитеты, регулярные отчёты);
- управление вовлечением:
  - поддержание поддержки союзников;
  - снижение сопротивления оппонентов;
  - разрешение конфликтов интересов (с приоритетом интересов заказчика, если нет иных жёстких ограничений).

<span style="color:red; font-weight:bold">ЗАПОМНИ:</span> стейкхолдеры могут «убить» отличный технически проект, если их интересы игнорировать, и наоборот – сильно помочь, если правильно с ними работать.
